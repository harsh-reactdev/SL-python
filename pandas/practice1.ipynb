```python
avgHigh = (df['Price'] > df['Price'].mean())
df.loc[avgHigh]
```

```python
avgPrices = df.groupby('Category')['Price'].mean()
```

```python
expensiveCategory = avgPrices.idxmax()
itsAvgValue = max(avgPrices)
print(expensiveCategory, '--->', itsAvgValue)
```

```python
stockVal = df.groupby('Category')['Stock'].count()
stockVal
```

```python
avgPrices = df.loc[df.groupby('Category')['Price'].idxmax()]
avgPrices
```

```python
category_brand_val = df.loc[df.groupby(['Category', 'Brand'])['Price'].idxmax()]
category_brand_val
```

```python
df.groupby('Category')['Stock'].count() # total number of products in each category
```

```python
df.groupby('Category')['Price'].mean() # average price of product in each category
```

```python
# total stock
df.groupby('Category')['Stock'].sum().sum()
```

```python
# in terms of brand, total number of products
df.groupby('Brand')['Stock'].count()
```

```python
# maximum priced product for each brand
df.groupby('Brand')['Price'].max()
```

```python
# minimum stock
df.groupby('Brand')['Stock'].min()
```

```python
# total sales computation
df['Total Sales'] = df['Price'] * df['Stock']
df
```

```python
# total sales by brand
totalSalesByBrand = df.groupby('Brand')['Total Sales'].sum()
totalSalesByBrand
```

```python
# brand with highest total sales
brand = totalSalesByBrand.idxmax()
value = max(totalSalesByBrand)
print(brand, '--->', value)
```

```python
# unique brands in each category
cat = df.groupby('Category')['Brand'].describe()
cat
```

```python
# highest average price 
catByPrice = df.groupby('Category')['Price'].mean()
sortedHighest = catByPrice.sort_values(ascending=False)

sortedHighest.head(3)
```
