FILTERING

```python
df['Employment'] == 'Employed' # returns a boolean series and not the intended rows that satisfies the condition
```

```python
df[df['WorkExp'] > 15] # returns the actual filtered results
```

```python
# another way to use filter
filt = df['WorkExp'] > 15
df.loc[filt, 'DevType']
```

```python
# logical ops in filters
filt2 = (df['Country'] == 'India') & (df['WorkExp'] > 25.0)
df.loc[filt2, ['DevType', 'CompTotal']]
```

```python
# to negate a filter condition, all we add is a tilde (~)
# we don't use ! for logical NOT, instead we use the tilde operator ~
filt2 = (df['Country'] == 'India') & (df['WorkExp'] > 25.0)
df.loc[~filt2, ['DevType', 'CompTotal']].sort_values('CompTotal', ascending=False)
```

```python
def conv(n):
    return round(n, 2)

indDf = df.loc[df['Country'] == 'India']
indDf.loc[df['WorkExp'] > 25.0, ['DevType', 'CompTotal']].sort_values('CompTotal')
indDf['CompTotal'].apply(conv)
```
