```python
import numpy as np
```

```python
# np.random.rand(shape) takes shape as argument
np.random.rand(2,3) # generates an ndArray with specified shape with random numbers between 0 and 1
```

```python
# randInt
print(np.random.randint(0, 10)) 
# generates a random integer between a specified range

print(np.random.randint(0, 10, 5)) 
# print(type(np.random.randint(0, 10, 5))) 
# generates an ndarray of 5 random integers between the specified range

print(np.random.randint(0, 10, (2, 3)))
# generates an ndarray of specified shape and fill it with random integers between the specified range
```

```python
# randn(shape) --> generates an ndarray of normal distribution for the given shape

# np.random.randn(3)
np.random.randn(2,3)
```

```python
# choice(options, size)

np.random.choice([10, 20, 30]) # choses one among the list of options
# np.random.choice([1,2,3,4], size=2) choses the specified number of options and returns an ndarray of chosen options
```

```python
print(np.random.seed(42))
```
