```python
import numpy as np
```

```python
car_brands = ["Toyota", "Honda", "Ford", "BMW", "Tesla"]
car_models = { 
    "Toyota": ["Camry", "Corolla", "RAV4"], 
    "Honda": ["Civic", "Accord", "CR-V"], 
    "Ford": ["Fiesta", "Focus", "Explorer"], 
    "BMW": ["X3", "X5", "3 Series"], 
    "Tesla": ["Model 3", "Model S", "Model X"]
}
car_prices = { "Camry": 25000, "Corolla": 20000, "RAV4": 30000, "Civic": 22000, "Accord": 26000, "CR-V": 28000, "Fiesta": 18000, "Focus": 20000, "Explorer": 35000, "X3": 42000, "X5": 55000, "3 Series": 48000, "Model 3": 40000, "Model S": 80000, "Model X": 90000 }
```

```python
recordType = [
    ('CustomerID', np.int32),
    ('CarBrand', 'U10'),
    ('CarModel', 'U15'),
    ('Units', np.int16),
    ('Amount', np.float64)
]
recordType
```

```python
# creating empty array of 100 records
maxRecords = 100

records = np.zeros(100, dtype=recordType)
records
```

```python
records[0]['CarBrand']
```

```python
records['CustomerID'] = np.arange(1001, 1101)
# records['CustomerID']

records['Units'] = np.random.randint(1, 6, 100)

records['CarBrand'] = np.random.choice(car_brands, size=100)
# records['CarBrand']
# records['Units']

carModels = [np.random.choice(car_models[carBrand]) for carBrand in records['CarBrand']]
records['CarModel'] = carModels
records
```

```python
# calculating amount

prices = [car_prices[carModel] for carModel in records['CarModel']]
records['Amount'] = prices * records['Units']
records
```

```python
# printing dimension of the so build np structured array
print(records.ndim)

# shape
print(records.shape)

# total sales
print(np.sum(records['Amount']))

# total units sold
print(np.sum(records['Units']))
```

```python
# records aggregator

def getRecords(brand, field, op):
    totalTeslaUnits = records[records['CarBrand'] == brand]
    res = op(totalTeslaUnits[field])
    print(res)
    return res
```

```python
# total revenue generated by BMW
getRecords('BMW', 'Amount', np.sum ) 

# total units sold, tesla only
getRecords('Tesla', 'Units', np.sum)

# average units sold, toyota only
getRecords('Toyota', 'Units', np.average)

# highest amount among tesla purchases
getRecords('Tesla', 'Amount', np.max)

# find brand having maximum total units sold
maxUnitsSold = 0
highestBrand = ''
for brand in car_brands:
    unitsSold = getRecords(brand, 'Units', np.sum)
    if unitsSold > maxUnitsSold:
        maxUnitsSold = unitsSold
        highestBrand = brand
print(highestBrand, maxUnitsSold)

# converting structured array into 2d numpy array
records2d = np.array(records, ndmin=2)
print(records2d)
```
