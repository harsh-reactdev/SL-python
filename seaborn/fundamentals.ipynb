```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme()
```

```python
data = [1, 2, 3, 4, 5, 6]
data2 = [2, 1, 4, 3, 6, 5]

plt.plot(data, data2)
plt.show()
```

```python
sns.lineplot(x="data", y="data2")
```

```python
df = pd.DataFrame({"data": data, "data2": data2})

sns.lineplot(x="data", y="data2", data=df)
```

```python
penguins = sns.load_dataset("penguins")
penguins.columns
```

```python
sns.lineplot(
    x="bill_depth_mm",  # x label
    y="bill_length_mm",  # y label
    data=penguins,  # data to use
    hue="sex",  # legend
    style="sex",
    palette="gnuplot",
)
plt.show()
```

```python
sns.scatterplot(
    data=penguins,
    x="bill_depth_mm",
    y="bill_length_mm",
    hue="sex",
    size="body_mass_g",
    style="sex",
    palette="magma",
)
```

```python
sns.relplot(
    data=penguins,
    x="bill_depth_mm",
    y="bill_length_mm",
    kind="scatter",
    hue="sex",
    size="body_mass_g",
    style="sex",
    palette="magma",
)
```

```python
sns.boxplot(
    data=penguins,
    x="island",
    y="species",
    hue="sex",
    palette="magma",
)
```
