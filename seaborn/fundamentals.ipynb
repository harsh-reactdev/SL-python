```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = [1, 2, 3, 4, 5, 6]
data2 = [2, 1, 4, 3, 6, 5]

plt.plot(data, data2)
plt.show()
```

```python
sns.lineplot(x='data', y='data2')
```

```python
df = pd.DataFrame({'data':data, 'data2':data2})

sns.lineplot(x='data', y='data2', data=df)
```

```python
penguins = sns.load_dataset('penguins')
```

```python
sns.lineplot(x='bill_depth_mm', y='bill_length_mm', data=penguins, hue='sex')
plt.show
```
