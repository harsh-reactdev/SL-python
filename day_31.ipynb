SETS and its METHODS

```python
s = {}
type(s) # an empty set returns dict when checked for its type

s = {1, 2, 3, 5}
type(s) # this returns as set

# hence to create an empty set we use the set constructor function
s = set()
type(s) # now this returns as set

# use for in loop to access set items
```

```python
# METHODS

s = {1, 2, 5, 6}
s2 = {3, 6, 7}

s.add(45)
s.pop()
s.discard(45)
s.union(s2)

del s

s2
```

```python
dict1 = {
    1: 'null',
    2: 'none',
    3: 'nada'
}

type(dict1.keys()) # returns as dict_keys, it is an iterable object

for key in dict1.keys():
    print(key, dict1[key])

# dic.keys()

dict1.values()
```
